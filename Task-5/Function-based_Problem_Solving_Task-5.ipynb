## Task-5 : Function Based Problem Solving

```python
# Problem 1 - Social Media Post Engagement Analyzer

def analyze_engagement(likes_list):
    total_likes = 0
    
    for likes in likes_list:
        total_likes += likes
    
    print("Total Likes:", total_likes)
    
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


analyze_engagement([200, 300, 250, 300])
```

```python
# Problem 2 - Medicine Stock Alert System

def medicine_stock_check(stock):
    print("Medicine Stock:", stock)
    
    if stock < 10:
        print("Status: Low Stock Alert")
    else:
        print("Status: Stock Sufficient")


medicine_stock_check(6)
```

```python
# Problem 3 - Rainfall Adequacy Checker

def rainfall_analysis(rainfall_data, required_level):
    total = 0
    
    for rain in rainfall_data:
        total += rain
    
    average = total / len(rainfall_data)
    
    print("Average Rainfall:", round(average, 2))
    
    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")


rainfall_analysis([60, 70, 80, 75], 70)
```

```python
# Problem 4 - Duplicate Account Detection

def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)
    
    if len(unique_users) < len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


detect_duplicate_accounts(["user1", "user2", "user1"])
```

```python
# Problem 5 - Appointment Eligibility Checker

def appointment_eligibility(age):
    print("Patient Age:", age)
    
    if age >= 18:
        print("Eligibility Status: Eligible")
    else:
        print("Eligibility Status: Not Eligible")


appointment_eligibility(21)
```

```python
# Problem 6 - Premium Crop Price Filter

def premium_crop_filter(prices):
    premium = []
    
    for price in prices:
        if price > 2000:
            premium.append(price)
    
    print("Premium Crops:", premium)


premium_crop_filter([1500, 2500, 3200, 1800])
```

```python
# Problem 7 - Application Health Checker

def check_application_health(error_count):
    print("Error Count:", error_count)
    
    if error_count == 0:
        print("System Status: Healthy")
    elif error_count <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")


check_application_health(7)
```

```python
# Problem 8 - Daily Transaction Limit Checker

def transaction_check(amount):
    daily_limit = 50000
    
    print("Transaction Amount:", amount)
    
    if amount <= daily_limit:
        print("Transaction Status: Approved")
    else:
        print("Transaction Status: Rejected")


transaction_check(60000)
```

```python
# Problem 9 - Student Attendance Eligibility System

def attendance_check(attendance_list):
    present_days = 0
    
    for day in attendance_list:
        if day == "P":
            present_days += 1
    
    percentage = (present_days / len(attendance_list)) * 100
    
    print("Attendance Percentage:", round(percentage, 2))
    
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")


attendance_check(["P", "P", "A", "P", "P"])
```

```python
# Problem 10 - Smart Electricity Bill Analyzer

def calculate_electricity_bill(units):
    bill = 0
    
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    # Usage Classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    
    return bill, usage


units_consumed = 250
bill_amount, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill:", bill_amount)
print("Usage Category:", usage_status)
```
