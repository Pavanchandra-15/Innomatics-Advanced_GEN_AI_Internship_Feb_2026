```python
# 1. User Login Check

def login_check(input_username, input_password):
    """
    Function to validate user login credentials.
    Returns login status message.
    """
    username = "admin"
    password = "1234"
    
    if input_username == username and input_password == password:
        return "Login Successful"
    else:
        return "Invalid Credentials"


# Test Cases
print(login_check("admin", "1234"))
print(login_check("user", "1111"))
```

```python
# 2. Pass / Fail Analyzer

def analyze_results(marks):
    """
    Analyzes pass and fail count from a list of marks.
    """
    pass_count = 0
    fail_count = 0
    
    for mark in marks:
        if mark >= 50:
            pass_count += 1
        else:
            fail_count += 1
    
    print("Total Students:", len(marks))
    print("Passed Students:", pass_count)
    print("Failed Students:", fail_count)

    # Pass percentage allows for easier data interpretation
    pass_percentage = (pass_count / len(marks)) * 100
    print("Pass Percentage:", round(pass_percentage, 2), "%")



marks = [45, 78, 90, 33, 60]
analyze_results(marks)
```

```python
# 3. Simple Data Cleaner

def clean_names(names):
    """
    Cleans and standardizes names.
    """
    cleaned = []
    
    for name in names:
        cleaned_name = name.strip().lower()
        cleaned.append(cleaned_name)
    
    return cleaned


names = [" Alice ", "bob", " CHARLIE "]
cleaned_names = clean_names(names)

print("Original Names:", names)
print("Cleaned Names:", cleaned_names)
print("Title Case Names:", [name.title() for name in cleaned_names])

```

```python
# 4. Message Length Analyzer

def analyze_messages(messages):
    """
    Analyzes message lengths and flags long messages.
    """
    for message in messages:
        length = len(message)
        print(f"Message: '{message}' | Length: {length}")
        
        if length > 10:
            print("⚠️ Flag: Message longer than 10 characters\n")
        else:
            print("✔️ Message length is acceptable\n")


messages = ["Hi", "Welcome to the platform", "OK"]
analyze_messages(messages)
```

```python
# 5. Error Message Detector

def count_errors(logs):
    """
    Counts the number of ERROR entries in logs.
    """
    error_count = logs.count("ERROR")
    
    print("Total Log Entries:", len(logs))
    print("Total ERROR Entries:", error_count)
    error_rate = (error_count / len(logs)) * 100
    print("Error Rate:", round(error_rate, 2), "%")
    
    return error_count


logs = ["INFO", "ERROR", "WARNING", "ERROR"]
count_errors(logs)
```
