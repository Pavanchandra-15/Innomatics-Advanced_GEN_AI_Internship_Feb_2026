```python
# Problem 1 - Mobile Recharge Validation System

def validate_recharge():
    valid_plans = [199, 299, 399, 599]
    
    while True:
        amount = int(input("Enter recharge amount: ₹"))
        
        if amount < 50:
            print("❌ Minimum recharge amount is ₹50.")
        
        elif amount not in valid_plans:
            print("❌ Invalid plan. Available plans:", valid_plans)
        
        else:
            print("✅ Recharge Successful!")
            break  # Exit loop if valid

if __name__ == "__main__":
    validate_recharge()
```

```python
# Problem 2 - Inventory Reorder Alert System

def check_inventory(stock):
    for product, quantity in stock.items():
        if quantity < 15:
            print(f"{product}: ⚠ Reorder Alert (Stock: {quantity})")
        else:
            print(f"{product}: ✔ Stock OK (Stock: {quantity})")


inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 20
}

check_inventory(inventory)
```

```python
# Problem 3 - Student Result Processing System

def student_result(marks):
    total = 0
    
    for mark in marks:
        total += mark
    
    average = total / len(marks)
    
    print("Average Marks:", round(average, 2))
    
    if average >= 50:
        return "Pass"
    else:
        return "Fail"


marks = [60, 45, 70, 55]
result = student_result(marks)
print("Result:", result)
```

```python
# Problem 4 - Cab Fare Estimator

def calculate_fare(distance, peak=False):
    base_fare = 50
    per_km_rate = 12
    
    fare = base_fare + (distance * per_km_rate)
    
    if peak:
        fare *= 1.25  # 25% extra
    
    return round(fare, 2)


def cab_system():
    while True:
        distance = float(input("Enter distance (km): "))
        peak_input = input("Is it peak hour? (yes/no): ").lower()
        
        peak = True if peak_input == "yes" else False
        
        fare = calculate_fare(distance, peak)
        print(f"Total Fare: ₹{fare}")
        
        retry = input("Do you want to calculate again? (yes/no): ").lower()
        if retry != "yes":
            break

if __name__ == "__main__":
    cab_system()
```

```python
# Problem 5 - Employee Attendance Eligibility

def attendance_eligibility(attendance):
    present_days = 0
    
    for day in attendance:
        if day == "P":
            present_days += 1
    
    percentage = (present_days / len(attendance)) * 100
    
    print("Attendance Percentage:", round(percentage, 2))
    
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


attendance = ["P", "P", "A", "P", "P", "A", "P"]
print("Status:", attendance_eligibility(attendance))
```

```python
# Problem 6 - Password Strength Checker

def check_password_strength(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"
    
    if len(password) < 8:
        return "Weak Password (Minimum 8 characters required)"
    
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True
    
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must include digit & special character)"


password = input("Enter your password: ")
print(check_password_strength(password))
```
