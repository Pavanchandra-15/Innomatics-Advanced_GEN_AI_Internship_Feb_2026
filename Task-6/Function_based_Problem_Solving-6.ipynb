```python
# Problem 1 - Smart Parking Lot Management System

def parking_system(capacity, logs):
    current_count = 0
    peak_usage = 0
    
    for log in logs:
        if log == "IN":
            current_count += 1
        elif log == "OUT" and current_count > 0:
            current_count -= 1
        
        # Track peak usage
        if current_count > peak_usage:
            peak_usage = current_count
    
    status = "Available" if current_count <= capacity else "Full"
    
    if current_count > capacity:
        alert = "⚠ Capacity Exceeded!"
    else:
        alert = "Within Capacity"
    
    return {
        "Currently Parked": current_count,
        "Peak Usage": peak_usage,
        "Parking Status": status,
        "Alert": alert
    }


result = parking_system(50, ["IN","IN","IN","OUT","IN","IN","OUT"])
print(result)
```

```python
# Problem 2 - Online Food Delivery Time Estimator

def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5  # 5 mins per km
    
    if traffic.lower() == "high":
        base_time *= 1.3
    elif traffic.lower() == "medium":
        base_time *= 1.1
    
    if weather.lower() == "rainy":
        base_time += 10
    elif weather.lower() == "storm":
        base_time += 20
    
    return round(base_time)


eta = estimate_delivery_time(8, "High", "Rainy")
print("Estimated Delivery Time:", eta, "minutes")
```

```python
# Problem 3 - Movie Theatre Seat Occupancy Analyzer

def theatre_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100
    
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"
    
    suggestion = "Open Additional Show" if occupancy >= 90 else "No Action Needed"
    
    return round(occupancy, 2), status, suggestion


occupancy, status, suggestion = theatre_occupancy(200, [1]*150)
print("Occupancy:", occupancy, "%")
print("Show Status:", status)
print("Suggestion:", suggestion)
```

```python
# Problem 4 - Cloud Server Load Classification

def classify_server_load(cpu_readings):
    avg_load = sum(cpu_readings) / len(cpu_readings)
    
    if avg_load < 50:
        status = "Normal"
    elif avg_load <= 80:
        status = "Warning"
    else:
        status = "Critical"
    
    return round(avg_load, 2), status


avg, status = classify_server_load([45, 60, 70, 85, 90])
print("Average CPU Load:", avg, "%")
print("Server Status:", status)
```

```python
# Problem 5 - Smart Classroom Resource Monitor

def resource_monitor(usage_dict):
    overused = []
    
    for resource, hours in usage_dict.items():
        if hours > 8:
            overused.append(resource)
    
    energy_alert = "Yes" if overused else "No"
    
    return overused, energy_alert


usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

overused, alert = resource_monitor(usage)
print("Overused Resources:", overused)
print("Energy Alert:", alert)
```

```python
# Problem 6 - Online Event Registration Capacity Controller

def event_registration(capacity, registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    
    return {
        "Confirmed Registrations": confirmed,
        "Waitlisted Users": waitlisted,
        "Registration Status": status
    }


result = event_registration(100, 105)
print(result)
```
