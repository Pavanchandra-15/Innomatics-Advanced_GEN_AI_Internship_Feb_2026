### 1. Employee Performance Bonus Eligibility

```python
# Problem 1 - Employee Performance Bonus Eligibility

def find_top_performers(employees):
    """
    Identifies employees with the highest performance score.
    Handles ties automatically.
    """
    highest_score = max(employees.values())
    
    top_employees = [
        name for name, score in employees.items()
        if score == highest_score
    ]
    
    print("Top Performers Eligible for Bonus:",
          ", ".join(top_employees),
          f"(Score: {highest_score})")
    
    return top_employees, highest_score


employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

find_top_performers(employees)
```

### 2. Search Query Keyword Analysis

```python
# Problem 2 - Search Query Keyword Analysis

import string

def keyword_analysis(query):
    """
    Cleans search query and returns keywords searched more than once.
    """
    query = query.lower()
    
    # Remove punctuation
    for char in string.punctuation:
        query = query.replace(char, "") 
    
    words = query.split() # Split the cleaned query into words
    
    frequency = {}
    
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1   # Count the frequency of each word
    
    repeated_keywords = {
        word: count for word, count in frequency.items() # List comprehension to filter words that appear more than once
        if count > 1
    }
    
    print(repeated_keywords)


keyword_analysis("Buy mobile phone buy phone online")
```

### 3. Sensor Data Validation

```python
# Problem 3 - Sensor Data Validation

def validate_sensor_data(sensor_readings):
    """
    Stores only even readings as valid data.
    Returns (hour_index, reading_value) pairs.
    """
    valid_readings = [
        (index, value)
        for index, value in enumerate(sensor_readings) # List comprehension to filter even readings and store hour index
        if value % 2 == 0
    ]
    
    print("Valid Sensor Readings (Hour, Value):")
    print(valid_readings)
    

sensor_readings = [3, 4, 7, 8, 10, 12, 5, 6, 9, 11, 14, 16, 18, 20, 22, 24, 26, 28, 30, 32, 34, 36, 38, 40]
validate_sensor_data(sensor_readings)
```

### 4. Email Domain Usage Analysis

```python
# Problem 4 - Email Domain Usage Analysis

def email_domain_analysis(emails):
    """
    Counts domain usage and calculates percentage.
    """
    domain_count = {}
    
    for email in emails:
        domain = email.split("@")[1] # split using "@" and take the domain part
        domain_count[domain] = domain_count.get(domain, 0) + 1 # Count the frequency of each domain
    
    total_users = len(emails)
    
    print("Email Domain Usage:")
    for domain, count in domain_count.items(): # items() to iterate through domain counts
        percentage = (count / total_users) * 100 
        print(f"{domain}: {round(percentage)}%")
    
    return domain_count


emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com", 
    "rohit@outlook.com",
    "priya@company.com",
    "david@company.com",
    "pallavi@outlook.com"
]

email_domain_analysis(emails)
```

### 5. Sales Spike Detection

```python
# Problem 5 - Sales Spike Detection

def detect_sales_spikes(sales):
    """
    Detects days where sales exceed 30% above average.
    """
    average_sales = sum(sales) / len(sales) # Calculate average sales
    threshold = average_sales * 1.3 # Calculate threshold for spike detection
    
    print("Average Sales:", round(average_sales, 2))
    print("Sales Spike Days:")
    
    for day, value in enumerate(sales, start=1): # Enumerate to get day index starting from 1
        if value > threshold:
            print(f"Day {day}: {value}")
    
    return average_sales


sales = [1200, 1500, 900, 2200, 1400, 3000, 1800, 2500, 1600, 2000, 2100, 1900, 2300, 2400, 2600, 2700, 2800, 2900, 3100, 3200, 3300,]
detect_sales_spikes(sales)
```

### 6. Duplicate User ID Detection

```python
# Problem 6 - Duplicate User ID Detection

def detect_duplicates(user_ids):
    """
    Detects duplicate user IDs and counts occurrences.
    """
    frequency = {}
    
    for user in user_ids:
        frequency[user] = frequency.get(user, 0) + 1 # Count the frequency of each user ID
    
    duplicates = {
        user: count for user, count in frequency.items() # List comprehension to filter user IDs that appear more than once
        if count > 1
    }
    
    for user, count in duplicates.items():
        print(f"{user} → {count} times")
    
    return duplicates


user_ids = ["user1", "user2", "user1", "user3", "user1", "user3", ]
detect_duplicates(user_ids)
```
