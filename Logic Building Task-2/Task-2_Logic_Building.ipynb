```python
# Problem 1 - Unique Words

def unique_words(sentence):
    words = sentence.lower().split() # I used this to separate the words based on whitespace and convert them to lowercase for case-insensitive comparison
    unique = set(words) # set data structure automatically removes duplicates, so we can easily get the unique words by converting the list of words to a set
    
    print("Unique words count:", len(unique))
    print("Unique words:", unique)
    
    return unique # This will return the set of unique words, but the order is not guaranteed since sets do not maintain order

sentence = "python is easy and python is powerful\nPython is great for data science"
unique_words(sentence)
```

```python
# Problem 2 - Highest Salary


def highest_salary(employees):
    highest_employee = max(employees, key=employees.get) # The max function is used to find the key (employee name) with the highest value (salary) in the employees dictionary. The key parameter is set to employees.get, which allows max to compare the values of the dictionary when determining the maximum.
    highest_value = employees[highest_employee]
    
    print(f"Highest Salary: {highest_employee} - {highest_value}") # Formatted print statement
    return highest_employee, highest_value

employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_salary(employees)
```

```python
# Problem 3 - Max & Min

def find_max_min(numbers):
    maximum = numbers[0] # Initialize maximum to the first element of the list
    minimum = numbers[0] # Initialize minimum to the first element of the list
    
    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    
    print("List:", numbers)
    print("Max:", maximum)
    print("Min:", minimum)
    
    return maximum, minimum

numbers = [45, 22, 89, 10, 66]
find_max_min(numbers)
```

```python
# Problem 4 - Products Above 1000

def count_expensive_products(prices, threshold=1000): # Default threshold is set to 1000, but it can be changed if needed
    count = 0
    for price in prices: # Iterate through each price in the list of prices
        if price > threshold:
            count += 1 # Increment the count if the price is above the threshold (1000)
            
    print(f"Products above {threshold}: {count}")
    return count

prices = [450, 1200, 899, 1500, 300]
count_expensive_products(prices)
```

```python
# Problem 5 - Attendance Percentage

def attendance_percentage(attendance):
    total_days = len(attendance)
    present_days = attendance.count("P") # The count method is used to count the number of occurrences of "P" in the attendance list, which gives us the number of days the student was present.
    
    percentage = (present_days / total_days) * 100
    
    print("Attendance Percentage:", round(percentage, 2)) # The round function is used to round the percentage to 2 decimal places for better readability.
    return percentage

attendance = ["P", "P", "A", "P", "P"]
attendance_percentage(attendance)
```

```python
# Problem 6 - Unique Phone Numbers

def unique_phone_numbers(numbers):
    unique_numbers = set(numbers) # Using a set to store unique phone numbers, as sets automatically handle duplicates
    print("Unique phone numbers:", unique_numbers)
    print("Count of unique phone numbers:", len(unique_numbers))
    print("Percentage of unique phone numbers:", (len(unique_numbers) / len(numbers)) * 100) # Calculate the percentage of unique phone numbers compared to the total list
    return unique_numbers

phone_numbers = [9876543210, 9123456789, 9876543210, 912345678, 9988776655, 923456789, 9019516016, 9980357289, 9988776655, 923456789, 9019516016, 9980357289]
unique_phone_numbers(phone_numbers)
```

```python
# Problem 7 - Character Frequency

def character_frequency(text):
    freq = {} # Initialize an empty dictionary to store the frequency of each character
    
    for char in text:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
    
    print(freq)
    return freq

character_frequency("python")
```

```python
# Problem 8 - List to Tuple
# This is a fundamental concept in any programming language (type conversion), as it allows us to convert a mutable data structure (list) into an immutable one (tuple). This can be useful in situations where we want to ensure that the data cannot be modified after it has been created.
def list_to_tuple(lst):
    converted = tuple(lst)
    print("Tuple:", converted)
    return converted

list_to_tuple([10, 20, 30])
```

```python
# Problem 9 - Check Key in Dictionary

def check_key(dictionary, key):
    if key in dictionary:
        print("Employee exists")
    else:
        print("Employee does not exist")

check_key(employees, "Ravi") # I used the 'employee' dictionary from Poblem 2.
check_key(employees, "Sita")
```

```python
# Problem 10 - Average Marks

def average_marks(marks):
    avg = sum(marks) / len(marks)
    print("Average Marks:", round(avg, 2)) # The round function is used to round the average marks to 2 decimal places for better readability.
 
average_marks([70, 80, 65, 75, 85, 91])
```
